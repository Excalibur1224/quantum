```python
import numpy as np
import matplotlib as plt
import time
```

```python
from qiskit import QuantumCircuit
```

```python
circuit = QuantumCircuit(3) #Creates template for program with 3 qubits
```

```python
circuit.h(0) #applies Hadamard gate on qubit 0

circuit.cx(0,1) #applies controlled not (cnot) on gate 1 determined by gate 0 after the hadamard opperation

circuit.x(0) #applies x gate on qubit 0
circuit.y(1) #applies y gate on qubit 1
circuit.z(2) #applies z gate on qubit 2

circuit.rz(np.pi/2,2) #applies half pi rotation along z on qubit 2

circuit.measure_all() #measures all qubits to collapse them into binary state
```

```python
print(circuit)
```

```python
circuit.draw(output='mpl')
```
